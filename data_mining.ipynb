# Tugas Data Mining: Eksplorasi Dataset Harga Rumah

Notebook ini membahas Data Loading, eksplorasi data, pembagian kolom kategorikal dan numerikal, serta pemeriksaan duplikat dan nilai null.

**Dataset:** House Prices – Advanced Regression Techniques (Kaggle). Unduh `train.csv` dari https://www.kaggle.com/competitions/house-prices-advanced-regression-techniques/data lalu unggah ke direktori kerja notebook.

## 1. Import library dan Data Loading

```python
from pathlib import Path
import pandas as pd

file_dataset = Path('train.csv')
if not file_dataset.exists():
    raise FileNotFoundError('File train.csv belum ditemukan. Unduh dari Kaggle dan unggah ke direktori kerja.')

df = pd.read_csv(file_dataset)
df.head()
```

## 2. Eksplorasi Data

```python
print('Ukuran dataset (baris, kolom):', df.shape)
print('Jumlah baris:', df.shape[0])
print('Jumlah kolom:', df.shape[1])
df.info()
```

## 3. Pembagian Data Kategorikal dan Numerikal

```python
data_kategorikal = df.select_dtypes(include=['object', 'category', 'string'])
data_numerikal = df.select_dtypes(include=['number'])

print('Jumlah kolom kategorikal:', data_kategorikal.shape[1])
print(data_kategorikal.columns.tolist())
print('\nJumlah kolom numerikal:', data_numerikal.shape[1])
print(data_numerikal.columns.tolist())
print('\nTipe data setiap kolom:')
display(df.dtypes.to_frame('tipe_data'))
```

### Contoh interpretasi
- **Kategorikal:** `MSZoning`, `Neighborhood`, `HouseStyle`, `RoofStyle`, `SaleCondition`.
- **Numerikal:** `LotArea`, `OverallQual`, `YearBuilt`, `GarageCars`, `SalePrice`.

Catatan: beberapa kolom numerik, seperti `MSSubClass`, menyimpan kode kategori sehingga perlu ditinjau berdasarkan makna kolom, bukan hanya tipe penyimpanannya.

## 4. Menghitung Data Duplikat dan Null

```python
jumlah_duplikat = int(df.duplicated().sum())
null_per_kolom = df.isnull().sum()
total_null = int(null_per_kolom.sum())

print('Jumlah baris duplikat:', jumlah_duplikat)
print('Total nilai null:', total_null)
print('\nKolom dengan nilai null:')
display(null_per_kolom[null_per_kolom > 0].sort_values(ascending=False).to_frame('jumlah_null'))
```

## 5. Statistik Deskriptif

```python
display(df.describe())
display(df.describe(include='object'))
```

## 6. Simpan Ringkasan Kualitas Data

```python
ringkasan = pd.DataFrame({
    'tipe_data': df.dtypes.astype(str),
    'jumlah_null': df.isnull().sum(),
    'jumlah_nilai_unik': df.nunique(dropna=True),
})
ringkasan.to_csv('ringkasan_kualitas_data.csv', index_label='nama_kolom')
print('File ringkasan_kualitas_data.csv berhasil dibuat.')
ringkasan.head()
```

## Kesimpulan
Tahap ini membaca dataset dengan Pandas, memeriksa ukuran dan tipe data, mengelompokkan kolom berdasarkan tipe penyimpanan, menghitung duplikat dan nilai null, serta menampilkan statistik deskriptif. Nilai duplikat dan null aktual harus diambil dari hasil menjalankan notebook pada file `train.csv` yang diunduh.
